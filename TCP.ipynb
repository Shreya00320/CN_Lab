```python

import socket
import os
import threading

HOST = '127.0.0.1'
PORT = 65432

def server():
    s = socket.socket()
    s.bind((HOST, PORT))
    s.listen(1)

    print("Server waiting...")

    conn, addr = s.accept()
    filename = conn.recv(1024).decode()

    if os.path.exists(filename):
        f = open(filename, 'r')
        data = f.read()
        f.close()
        conn.send(data.encode())
    else:
        conn.send("File not found".encode())

    conn.close()
    s.close()

thread = threading.Thread(target=server)
thread.start()
```
