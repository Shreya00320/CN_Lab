```python

capacity = int(input("Enter bucket size: "))
rate = int(input("Enter processing rate: "))
n = int(input("Enter number of seconds: "))

packets = []
for i in range(n):
    packets.append(int(input("Packets at second " + str(i+1) + ": ")))

bucket = 0

print("\nSec\tReceived\tSent\tLeft\tDropped")

for i in range(n):
    bucket += packets[i]
    dropped = max(0, bucket - capacity)
    bucket = min(bucket, capacity)

    sent = min(bucket, rate)
    bucket -= sent

    print(i+1, "\t", packets[i], "\t\t", sent,
          "\t", bucket, "\t", dropped)

while bucket > 0:
    sent = min(bucket, rate)
    bucket -= sent
    print("Drain\t0\t\t", sent, "\t", bucket, "\t0")
```
